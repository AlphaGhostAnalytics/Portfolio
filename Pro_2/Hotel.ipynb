```python
df = pd.read_csv('hotel_bookings.csv')
df.info()
```

```python
# Filling missing values for agent, company and country and assigning the right data types to them
df['agent'] = df['agent'].fillna(0).astype(int).astype(str)
df['company'] = df['company'].fillna(0).astype(int).astype(str)
df['country'] = df['country'].fillna('Unknown')
```

```python
df['children'] = df['country'].fillna(0)
```
