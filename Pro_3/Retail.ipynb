```python
df = pd.read_csv('Retail.csv')
```

```python
df.columns
```

```python
df.drop(columns = 'Unnamed: 0', inplace = True)
```

```python
df.columns
```

```python
df.info()
```

```python
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
df['Revenue'] = df['Quantity'] * df['Price']
# Add Cancelled (invoice starts with 'C')
df['Invoice'] = df['Invoice'].astype(str)
df['Cancelled'] = df['Invoice'].str.startswith('C')

# Add Products (excluding service/admin stock codes)
non_product_codes = {
    'POST', 'DOT', 'M', 'm', 'D', 'S', 'ADJUST', 'AMAZONFEE',
    'CRUK', 'B', 'GIFT', 'BANK CHARGES', 'PADS'
}
df['StockCode'] = df['StockCode'].astype(str)
df['Products'] = ~df['StockCode'].isin(non_product_codes)
```

```python
df.columns
```

```python
df.to_csv("Retail.csv", index = False)
print('DONE')
```
